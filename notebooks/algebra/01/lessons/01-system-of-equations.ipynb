## Plain English


A system of linear equations consists of two or more linear equations made up of two or more variables such that all equations in the system are considered simultaneously. To find the unique solution to a system of linear equations, we must find a numerical value for each variable in the system that will satisfy all equations in the system at the same time. (OpenStax)

# Types of Systems of Linear Equations

* **Consistent system**

  * Has **at least one solution**.

* **Independent system**

  * A consistent system with **exactly one solution**.
  * The equations represent two lines with **different slopes**.
  * The lines intersect at **one point** in the plane.

* **Dependent system**

  * A consistent system with **infinitely many solutions**.
  * The equations have the **same slope** and the **same $y$-intercept**.
  * The lines **coincide**, meaning they represent the same line.
  * Every point on the line is a solution to the system.

* **Inconsistent system**

  * Has **no solution**.
  * The equations represent two **parallel lines**.
  * The lines have the **same slope** but **different $y$-intercepts**.
  * Since the lines never intersect, there are **no points common to both lines**.
