## Exercise 1

As the $x$-values get closer and closer to 2 from both the left and the right (without ever reaching 2), the corresponding $y$-values of $f(x)$ get closer and closer to 5.

But even though the $y$-values approach 5 as $x$ gets closer to 2, the actual function value at that exact single point, $f(2)$, can be 3.

For example, the piecewise function

$$
f(x) = \begin{cases} x+3, & x \neq 2 \\ 3, & x = 2 \end{cases}
$$

shows that $\lim_{x \to 2} f(x) = 5$ and $f(2) = 3$ can both be true.

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Define the function for x != 2 (the line y = x + 3)
# ------------------------------------------------------------
# We create a dense array of x-values from 0 to 4
x_vals = np.linspace(0, 4, 1000)

# Calculate y = x + 3 for ALL these points
y_vals = x_vals + 3

# ------------------------------------------------------------
# 2. Find the index where x is exactly 2
# ------------------------------------------------------------
# np.linspace creates many points, but probably none are *exactly* 2.
# We find the closest index to x=2 so we can remove it to create a "hole".
index_at_2 = np.argmin(np.abs(x_vals - 2))

# Create a mask (True/False array) that is True everywhere EXCEPT at x=2
mask = np.ones(len(x_vals), dtype=bool)
mask[index_at_2] = False

# Apply the mask: this removes that single point (2,5) from the line
x_line = x_vals[mask]
y_line = y_vals[mask]

# ------------------------------------------------------------
# 3. Create the plot
# ------------------------------------------------------------
plt.figure(figsize=(8, 6))

# --- Plot the main line with a HOLE at x=2 ---
# We use 'bo-' which means blue circles connected by a line.
# Since we removed the point at x=2, the line will visually break.
plt.plot(x_line, y_line, 'bo-', label='f(x) = x + 3, for x ≠ 2', markersize=4)

# --- Explicitly draw the HOLE at (2,5) as an open circle ---
# This makes the "missing point" very obvious.
plt.plot(2, 5, 'wo', markerfacecolor='white', markersize=12, markeredgecolor='blue', label='Hole (limit = 5)')

# --- Explicitly draw the ACTUAL point at (2,3) ---
# This is the isolated dot that makes f(2) = 3.
plt.plot(2, 3, 'ro', markersize=10, label='f(2) = 3 (isolated point)')

# ------------------------------------------------------------
# 4. Decorate the graph
# ------------------------------------------------------------
plt.title('Limit vs. Function Value: A Classic Calculus Example', fontsize=14)
plt.xlabel('x values', fontsize=12)
plt.ylabel('f(x) values', fontsize=12)
plt.axhline(y=5, color='gray', linestyle='--', linewidth=0.8, alpha=0.5)  # Dashed line at y=5
plt.axvline(x=2, color='gray', linestyle='--', linewidth=0.8, alpha=0.5)  # Dashed line at x=2
plt.grid(True, alpha=0.3)
plt.legend(loc='upper left')
plt.xlim(0, 4)
plt.ylim(2, 6)

# Annotate the points so they are easy to read
plt.text(2.1, 4.8, 'Limit approaches 5', fontsize=10, color='blue')
plt.text(2.1, 2.8, 'But f(2) is actually 3', fontsize=10, color='red')

plt.tight_layout()
plt.show()
```

## Exercise 19

Estimate $\lim_{x \to 3} \dfrac{x^2 - 3x}{x^2 - 9}$ numerically. The table below suggests the limit is $\tfrac{1}{2}$.

```python
def f(x):
    num = x**2 - 3*x
    den = x**2 - 9
    return num / den

x_left = [2.9, 2.95, 2.99, 2.999, 2.9999]
x_right = [3.1, 3.05, 3.01, 3.001, 3.0001]

y_left = [f(a) for a in x_left]
y_right = [f(a) for a in x_right]

print('Approaching from the left:')
for x, y in zip(x_left, y_left):
    print(f'x: {x} --> f(x): {y:.6f}')

print('\nApproaching from the right:')
for x, y in zip(x_right, y_right):
    print(f'x: {x} --> f(x): {y:.6f}')
```

## Exercise 21

Estimate $\lim_{x \to 0} \dfrac{\sin x}{x + \tan x}$ numerically. The table below suggests the limit is $\tfrac{1}{2}$.

```python
import math

def f(x):
    num = math.sin(x)
    den = x + math.tan(x)
    return num / den

# Ordered from farthest to closest to zero in both directions
x_neg = [-1, -0.5, -0.2, -0.1, -0.05, -0.01]
x_pos = [1, 0.5, 0.2, 0.1, 0.05, 0.01]

y_neg = [f(x) for x in x_neg]
y_pos = [f(x) for x in x_pos]

print('Approaching from the left (negative):')
for x, y in zip(x_neg, y_neg):
    print(f'x: {x} --> f(x): {y:.6f}')

print('\nApproaching from the right (positive):')
for x, y in zip(x_pos, y_pos):
    print(f'x: {x} --> f(x): {y:.6f}')
```

## Exercise 23

Estimate $\lim_{\theta \to 0} \dfrac{\sin(3\theta)}{\tan(2\theta)}$ numerically. The table below suggests the limit is $\tfrac{3}{2}$.

```python
import math

def f23(theta):
    return math.sin(3 * theta) / math.tan(2 * theta)

th_esq = [-0.1, -0.01, -0.001]
th_dir = [0.1, 0.01, 0.001]

print('Approaching from the left:')
for x in th_esq:
    print(f'theta: {x} --> f(theta): {f23(x):.6f}')

print('\nApproaching from the right:')
for x in th_dir:
    print(f'theta: {x} --> f(theta): {f23(x):.6f}')
```
