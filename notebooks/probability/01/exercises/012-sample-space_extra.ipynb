1) Seja X a variável aleatória correspondente à soma dos pontos obtidos no lançamento de dois dados. Determine:

a) a distribuição de probabilidades de X;

b) P(3 ≤ X ≤ 10);

c) P(X > 7);

d) P(X ≤ 5).



```python
faces = range(1, 7)
espaco_amostral = []

for d1 in faces:
  for d2 in faces:
    espaco_amostral.append((d1, d2))

espaco_amostral
```

```python
somas = []

for a in espaco_amostral:
  resultado = a[0] + a[1]
  somas.append(resultado)
```

```python
from collections import Counter

# Conta quantas vezes cada soma ocorreu
frequencias = Counter(somas)
total = len(espaco_amostral)
frequencias
```

```python
print(f"{'X':<5} | {'Frequência':<10} | {'Probabilidade':<15}")
print("-" * 35)

for x in sorted(frequencias):
  freq = frequencias[x]
  prob = freq / total
  print(f"{x:<5} | {freq:<10} | {prob:.4f} ({freq}/{total})")
```

2) Uma variável aleatória tem a distribuição de probabilidade dada pela seguinte fórmula:

P(xi) = K/x   para x = 1, 3, 5, 7.

a) Determinar K.

b) Calcular P(2 ≤ X ≤ 6).

P(x1) + P(x2) ... P(x7) = 1

```python
total = []
x = [1,3,5,7]
```

```python
valores_x = [1, 3, 5, 7]

soma_inversos = 0
for x in valores_x:
  soma_inversos += 1 / x

K = 1 / soma_inversos
print("Valor de K:", K)
```

```python
valores_x = [1, 3, 5, 7]

prob_intervalo = 0

for x in valores_x:

  if 2 <= x <= 6:
    prob_atual = K / x

    prob_intervalo += prob_atual

print("b) P(2 <= X <= 6) =", prob_intervalo)
```
