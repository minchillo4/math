## 1. Sum of Two Dice

Let $X$ be the random variable corresponding to the sum of the points obtained when rolling two fair dice. Determine:

(a) The probability distribution of $X$

(b) $P(3 \le X \le 10)$

(c) $P(X > 7)$

(d) $P(X \le 5)$

### (a) Probability Distribution of $X$

Build the sample space by listing all ordered pairs $(d_1, d_2)$:

```python
faces = range(1, 7)
sample_space = [(d1, d2) for d1 in faces for d2 in faces]
sample_space
```

```python
sums = [d1 + d2 for d1, d2 in sample_space]
frequencies = Counter(sums)
total = len(sample_space)
frequencies
```

```python
print(f"{'X':<5} | {'Frequency':<10} | {'Probability':<15}")
print("-" * 35)

for x in sorted(frequencies):
    freq = frequencies[x]
    prob = freq / total
    print(f"{x:<5} | {freq:<10} | {prob:.4f} ({freq}/{total})")
```

### (b) $P(3 \le X \le 10)$

```python
prob_b = sum(frequencies[x] / total for x in frequencies if 3 <= x <= 10)
print(f"P(3 <= X <= 10) = {prob_b:.4f}")
```

### (c) $P(X > 7)$

```python
prob_c = sum(frequencies[x] / total for x in frequencies if x > 7)
print(f"P(X > 7) = {prob_c:.4f}")
```

### (d) $P(X \le 5)$

```python
prob_d = sum(frequencies[x] / total for x in frequencies if x <= 5)
print(f"P(X <= 5) = {prob_d:.4f}")
```

## 2. Unknown Constant K

A random variable has the probability distribution:

$$P(X = x) = \frac{K}{x}, \quad x \in \{1, 3, 5, 7\}$$

(a) Determine $K$.

(b) Calculate $P(2 \le X \le 6)$.

### (a) Determine $K$

Since probabilities must sum to 1:

$$\sum_{x \in \{1,3,5,7\}} \frac{K}{x} = 1 \implies K \left(\frac{1}{1} + \frac{1}{3} + \frac{1}{5} + \frac{1}{7}\right) = 1$$

```python
x_values = [1, 3, 5, 7]
sum_reciprocals = sum(1 / x for x in x_values)
K = 1 / sum_reciprocals
print(f"K = {K}")
```

### (b) Calculate $P(2 \le X \le 6)$

Among the support values $\{1, 3, 5, 7\}$, only $x = 3$ and $x = 5$ fall in the interval $[2, 6]$.

```python
prob_2b = sum(K / x for x in x_values if 2 <= x <= 6)
print(f"P(2 <= X <= 6) = {prob_2b:.6f}")
```

## 3. Sales Contract Probability

A salesperson calculates that each contact results in a sale with probability 20%. On a given day, she contacts two potential customers. Construct the probability distribution table for the random variable $Y$: number of customers who sign a sales contract.

```python
p_sale = 0.2
p_no_sale = 1 - p_sale

# Sample space: {SS, SN, NS, NN} where S=sale, N=no sale
y_0 = p_no_sale ** 2          # P(Y=0) = NN
y_1 = 2 * (p_sale * p_no_sale)  # P(Y=1) = SN + NS
y_2 = p_sale ** 2             # P(Y=2) = SS

print(f"{'Y':<5} | {'Probability':<15}")
print("-" * 25)
print(f"{0:<5} | {y_0:.4f}")
print(f"{1:<5} | {y_1:.4f}")
print(f"{2:<5} | {y_2:.4f}")
print(f"\nSum: {y_0 + y_1 + y_2:.4f}")
```

## 4. Probability Distribution — Finding $P(X=3)$

A random variable $X$ has the following distribution:

| $x$ | $P(x)$ |
|-----|--------|
| 1   | 0.20   |
| 2   | 0.25   |
| 3   | ?      |
| 4   | 0.30   |
| 5   | 0.10   |

(a) Find $P(X = 3)$.

(b) Find the expected value, variance, and standard deviation.

### (a) Find $P(X = 3)$

Since all probabilities must sum to 1:

```python
p_3 = 1 - (0.20 + 0.25 + 0.30 + 0.10)
print(f"P(X=3) = {p_3:.4f}")
```

### (b) Expected Value, Variance, and Standard Deviation

```python
x_vals = [1, 2, 3, 4, 5]
p_vals = [0.20, 0.25, p_3, 0.30, 0.10]

expected_value = sum(x * p for x, p in zip(x_vals, p_vals))
print(f"E[X] = {expected_value:.4f}")
```

```python
variance = sum((x - expected_value) ** 2 * p for x, p in zip(x_vals, p_vals))
print(f"Var(X) = {variance:.4f}")
```

```python
std_dev = math.sqrt(variance)
print(f"Std(X) = {std_dev:.6f}")
```

## 5. Geometric Distribution — Infinite Tail

Consider a geometric distribution with $p = 0.2$:

$$P(X = r) = 0.8^{\,r-1} \cdot 0.2, \quad r = 1, 2, 3, \dots$$

(a) Compute $P(X = r)$ for $r = 1, 2, 3, 4, 5$.

(b) What is $P(X > 5)$?

### (a) Probabilities for $r = 1, \dots, 5$

```python
for r in range(1, 6):
    p_r = 0.8 ** (r - 1) * 0.2
    print(f"P(X={r}) = {p_r:.5f}")
```

```python
cumulative = sum(0.8 ** (r - 1) * 0.2 for r in range(1, 6))
print(f"\nSum of first 5 terms: {cumulative}")
```

### (b) $P(X > 5)$

The sum of all probabilities for $x > 5$ exists and represents the "remainder" needed for the total distribution to equal 1 (exactly $0.00032$, complementing the $0.99968$ from the first five terms). These probabilities seem to "disappear" only because they become extremely small decimal values, tending toward zero.

```python
print(f"P(X > 5) = {1 - cumulative:.5f}")
```

## 6. Call Center Distribution

A call center records the number of calls needed to resolve a customer issue. The distribution is:

| Calls ($x$) | $P(X = x)$ |
|------------|------------|
| 0          | 0.55       |
| 1          | 0.25       |
| 2          | 0.10       |
| 3          | 0.04       |
| 4          | 0.04       |
| 5          | 0.02       |

(a) Find $P(1 \le X \le 4)$.

(b) Find $P(X > 1)$.

```python
n_calls = [0, 1, 2, 3, 4, 5]
probs = [0.55, 0.25, 0.10, 0.04, 0.04, 0.02]

for call, prob in zip(n_calls, probs):
    print(f"Calls: {call} -> P = {prob * 100:.0f}%")

print(f"\nSanity check (sum): {sum(probs)}")

p_1_to_4 = sum(prob for call, prob in zip(n_calls, probs) if 1 <= call <= 4)
p_gt_1 = sum(prob for call, prob in zip(n_calls, probs) if call > 1)

print(f"P(1 <= X <= 4) = {p_1_to_4:.2f}")
print(f"P(X > 1) = {p_gt_1:.2f}")
```

## 7. IBM Computer Users

According to a Data Journal survey, 70% of office workers use IBM computers. If two office workers are selected at random, find the probability distribution of $X$: number of IBM computer users. Calculate the mean and standard deviation.

```python
p_ibm = 0.7
p_not = 1 - p_ibm

# Sample space: {II, IN, NI, NN} where I=IBM, N=not IBM
p0 = p_not ** 2           # P(X=0) = NN
p1 = 2 * (p_ibm * p_not)  # P(X=1) = IN + NI
p2 = p_ibm ** 2           # P(X=2) = II

print(f"P(X=0) = {p0:.2f}")
print(f"P(X=1) = {p1:.2f}")
print(f"P(X=2) = {p2:.2f}")
print(f"Sum: {p0 + p1 + p2:.4f}")

expected_value = 0 * p0 + 1 * p1 + 2 * p2
print(f"\nE[X] = {expected_value:.4f}")

variance = (0 - expected_value) ** 2 * p0 + \
           (1 - expected_value) ** 2 * p1 + \
           (2 - expected_value) ** 2 * p2
print(f"Var(X) = {variance:.4f}")

std_dev = math.sqrt(variance)
print(f"Std(X) = {std_dev:.6f}")
```

## 8. Hurricane Categories

The probability distribution of hurricane categories that hit the USA is given by:

| Category | $P(x)$ |
|----------|--------|
| 1        | 0.399  |
| 2        | 0.264  |
| 3        | 0.260  |
| 4        | 0.066  |
| 5        | 0.011  |

Find the expected value, variance, and standard deviation.

```python
categories = [1, 2, 3, 4, 5]
probs_8 = [0.399, 0.264, 0.260, 0.066, 0.011]

plt.figure(figsize=(8, 5))
plt.bar(categories, probs_8, color='#d3d3d3', edgecolor='black', width=1.0, align='center')
plt.title('Hurricane Categories Hitting the USA', fontsize=12, fontweight='bold')
plt.xlabel('Category', fontsize=10)
plt.ylabel('P(x)', fontsize=10)
plt.xticks(categories)
plt.ylim(0, 0.45)

for cat, prob in zip(categories, probs_8):
    plt.text(cat, prob + 0.01, f"{prob:.3f}", ha="center", va="bottom", fontsize=9)

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
expected_value_8 = sum(cat * prob for cat, prob in zip(categories, probs_8))
print(f"E[X] = {expected_value_8:.4f}")

variance_8 = sum((cat - expected_value_8) ** 2 * prob for cat, prob in zip(categories, probs_8))
print(f"Var(X) = {variance_8:.6f}")

std_dev_8 = math.sqrt(variance_8)
print(f"Std(X) = {std_dev_8:.6f}")
```

## 10. American Roulette — Expected Value

In American roulette there are 38 numbers: 00, 0, 1, 2, 3, …, 36, marked on equally sized spaces. If a player bets \$1 on a number and wins, they keep the \$1 and receive an additional \$35. Otherwise, they lose the \$1.

Let $X$ be the profit from one round. Determine the average amount of money per game that this player can expect to lose.

```python
prob_win = 1 / 38
prob_lose = 37 / 38
profit_win = 35
profit_lose = -1

expected_value = (prob_win * profit_win) + (prob_lose * profit_lose)
print(f"Expected profit per game: {expected_value:.4f}")
print(f"The player expects to lose approximately ${abs(expected_value):.4f} per game.")
```
